```python
import pandas as pd, shutil
f = "anotasi/anotasi_Rani.xlsx"
d = pd.read_excel(f, sheet_name="Anotasi", dtype=str).fillna("")
terisi = d[d["label"].str.strip() != ""]
print(f"{len(d)} baris, {len(terisi)} terisi")
assert len(terisi) == 300, "Jumlah terisi bukan 300, berhenti dulu dan kabari aku"
shutil.copy(f, "data/cadangan_anotasi_Rani_700baris.xlsx")   # cadangan, di luar folder anotasi/
terisi.to_excel(f, sheet_name="Anotasi", index=False)
print("Disimpan ulang:", len(terisi), "baris")
```
